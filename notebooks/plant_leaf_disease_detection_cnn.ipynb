## 1. Environment Setup & Dependency Ingestion
Import analytical, statistical, and Machine Learning packages.

```python
import os
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
from sklearn.metrics import classification_report, confusion_matrix

# Visual configuration
plt.style.use('seaborn-v0_8-whitegrid')
print('Environment initialized successfully!')
```

## 2. Dataset Loading & Exploratory Data Analysis (EDA)
Load the Kaggle metadata containing extracted morphological, texture, and chlorosis features.

```python
df = pd.read_csv('../data/dataset_metadata.csv')
print(f'Total Samples Loaded: {len(df):,}')
display(df.head(8))

# Class distribution
plt.figure(figsize=(10, 4))
sns.countplot(data=df, x='label', palette='viridis', order=df['label'].value_counts().index)
plt.title('Distribution of Pathology Classes in Dataset', fontweight='bold')
plt.xlabel('Disease Category')
plt.ylabel('Number of Samples')
plt.xticks(rotation=25)
plt.tight_layout()
plt.show()
```

## 3. Feature Distribution & Pathology Biomarker Analysis
Analyze how greenness index, chlorosis index, lesion area ratio, and GLCM texture contrast separate disease categories.

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.boxplot(data=df, x='label', y='chlorosis_index', ax=axes[0], palette='Set2')
axes[0].set_title('Chlorosis (Yellowing) Index across Classes', fontweight='bold')
axes[0].tick_params(axis='x', rotation=30)

sns.boxplot(data=df, x='label', y='lesion_area_ratio', ax=axes[1], palette='magma')
axes[1].set_title('Lesion Necrosis Area Ratio across Classes', fontweight='bold')
axes[1].tick_params(axis='x', rotation=30)

plt.tight_layout()
plt.show()
```

## 4. Model Benchmarks & Diagnostic Evaluation
Evaluate trained classifier model metrics, confusion matrix, and multi-class classification report.

```python
model = joblib.load('../models/plant_leaf_disease_model.joblib')
with open('../reports/metrics.json') as f:
    metrics = json.load(f)

print('=' * 60)
print('      PLANT LEAF PATHOLOGY CLASSIFIER BENCHMARK METRICS      ')
print('=' * 60)
print(f"Test Accuracy  : {metrics['Test_Accuracy_Pct']}%")
print(f"Macro F1-Score : {metrics['Macro_F1_Score']}")
print(f"Macro Precision: {metrics['Macro_Precision']}")
print(f"Macro Recall   : {metrics['Macro_Recall']}")
print('=' * 60)
```

## 5. Normalized Confusion Matrix Heatmap

```python
from PIL import Image
img = Image.open('../reports/confusion_matrix.png')
plt.figure(figsize=(9, 7))
plt.imshow(img)
plt.axis('off')
plt.title('Evaluated Test Confusion Matrix', fontweight='bold', fontsize=14)
plt.show()
```

## 6. Real-Time Inference & Treatment Prescription
Demonstrate automated diagnostic scoring and agronomical intervention recommendations.

```python
sample_unseen = pd.DataFrame([{
    'mean_greenness_index': 0.46,
    'chlorosis_index': 0.36,
    'lesion_area_ratio': 0.34,
    'glcm_contrast': 40.2
}])

predicted_class = model.predict(sample_unseen)[0]
probabilities = model.predict_proba(sample_unseen)[0]
confidence = max(probabilities) * 100

print(f"🌾 Diagnostic Result: {predicted_class.replace('_', ' ')}")
print(f"🎯 Confidence Score: {confidence:.2f}%")
```
